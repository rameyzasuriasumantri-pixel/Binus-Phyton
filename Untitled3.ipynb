```python
def write_biodata(filename):
    name = input("Name: ")
    age = input("Age: ")
    address = input("Address: ")
    email = input("Email: ")

    with open(filename, "w") as file:
        file.write(f"Name: {name}\n")
        file.write(f"Age: {age}\n")
        file.write(f"Address: {address}\n")
        file.write(f"Email: {email}\n")

    print("\nFile created successfully!\n")
```

```python
def read_biodata(filename):
    with open(filename, "r") as file:
        text = file.read()
    print(text)
```

```python
filename = "Biodata.txt"
write_biodata(filename)
read_biodata(filename)
```

```python
def create_file(filename):
    text = input("Enter text: ")
    with open(filename, "w") as file:
        file.write(text + "\n")
    print("File created!")


def append_file(filename):
    text = input("Enter text to add: ")
    with open(filename, "a") as file:
        file.write(text + "\n")
    print("Text added!")
```

```python
def read_file(filename):
    with open(filename, "r") as file:
        print(file.read())
```

```python
while True:
    print("1. Create file")
    print("2. Read file")
    print("3. Append file")
    print("4. Close")
    choice = input("Choose (1/2/3/4): ")

    if choice == "4":
        print("Goodbye!")
        break

    filename = input("File name: ")

    try:
        if choice == "1":
            create_file(filename)
        elif choice == "2":
            read_file(filename)
        elif choice == "3":
            append_file(filename)
        else:
            print("Invalid choice")
    except OSError:
        print("Invalid file name or file cannot be opened.")
    finally:
        print("Done.\n")
```
